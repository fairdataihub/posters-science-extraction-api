# PDF to Image Conversion Test

This notebook tests the PDF to image conversion functionality of the Poster Extraction API.
Conversion is performed using the `pymupdf`. If multiple pages are present in the PDF, only the first page is converted to an image.

```python
import pymupdf 

source_pdf = "/Users/sanjay/Developer/posters-science-extraction-api/example_posters/42.pdf"
output_img = "/Users/sanjay/Developer/posters-science-extraction-api/thumbail_generation/42.jpeg"

# generate the thumbnail image parent folder if it doesn't exist
import os
os.makedirs(os.path.dirname(output_img), exist_ok=True)
```

```python
THUMBNAIL_MAX_WIDTH = 1200  # px — enough for display, keeps file size small
THUMBNAIL_JPEG_QUALITY = 75  # 1-100; 75 is a good balance for thumbnails


def generate_thumbnail(pdf_path, output_path, page_num=0):
    """
    Convert a single PDF page to a compressed JPEG thumbnail.

    The pixmap is scaled so the width does not exceed THUMBNAIL_MAX_WIDTH,
    then encoded at THUMBNAIL_JPEG_QUALITY to keep the file size small.

    Args:
        pdf_path: Path to the PDF file
        output_path: Path for the output JPEG image
        page_num: Page number to convert (0-indexed)
    """
    doc = pymupdf.open(pdf_path)
    try:
        page = doc[page_num]
        # Scale so width <= THUMBNAIL_MAX_WIDTH (height scales proportionally)
        natural_width = page.rect.width  # points at 72 dpi
        scale = min(1.0, THUMBNAIL_MAX_WIDTH / natural_width)
        matrix = pymupdf.Matrix(scale, scale)
        pix = page.get_pixmap(matrix=matrix)
        pix.save(output_path, jpg_quality=THUMBNAIL_JPEG_QUALITY)
        print(f"Saved: {output_path}  ({pix.width}x{pix.height}px, scale={scale:.2f}, quality={THUMBNAIL_JPEG_QUALITY})")
    finally:
        doc.close()


generate_thumbnail(source_pdf, output_img)
```
